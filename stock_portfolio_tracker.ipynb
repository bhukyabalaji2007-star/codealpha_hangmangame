```python
# CodeAlpha Task 2: Stock Portfolio Tracker

# Hardcoded stock prices
stock_prices = {
    "AAPL": 180,
    "TSLA": 250,
    "GOOGL": 175,
    "MSFT": 420,
    "AMZN": 185
}

portfolio = {}
total_investment = 0

print("======================================")
print("       STOCK PORTFOLIO TRACKER")
print("======================================")

# Display available stocks
print("\nAvailable Stocks:")
for stock, price in stock_prices.items():
    print(f"{stock} : ${price}")

# Number of stocks to enter
number_of_stocks = int(input("\nHow many stocks do you want to add? "))

# Get user input
for i in range(number_of_stocks):
    stock = input(f"\nEnter stock name {i + 1}: ").upper()

    if stock in stock_prices:
        quantity = int(input(f"Enter quantity of {stock}: "))

        investment = stock_prices[stock] * quantity
        portfolio[stock] = {
            "quantity": quantity,
            "price": stock_prices[stock],
            "investment": investment
        }

        total_investment += investment

        print(f"{stock} added successfully!")
    else:
        print("❌ Stock not available in the list.")

# Display portfolio
print("\n======================================")
print("          YOUR PORTFOLIO")
print("======================================")

for stock, details in portfolio.items():
    print(
        f"{stock} | "
        f"Quantity: {details['quantity']} | "
        f"Price: ${details['price']} | "
        f"Investment: ${details['investment']}"
    )

print("--------------------------------------")
print(f"Total Investment: ${total_investment}")
print("======================================")

# Save result to a text file
save = input("\nDo you want to save the result? (yes/no): ").lower()

if save == "yes":
    with open("portfolio.txt", "w") as file:
        file.write("STOCK PORTFOLIO TRACKER\n")
        file.write("========================\n")

        for stock, details in portfolio.items():
            file.write(
                f"{stock} | Quantity: {details['quantity']} | "
                f"Price: ${details['price']} | "
                f"Investment: ${details['investment']}\n"
            )

        file.write(f"\nTotal Investment: ${total_investment}")

    print("✅ Portfolio saved successfully as portfolio.txt")

else:
    print("Portfolio was not saved.")

print("\nThank you for using Stock Portfolio Tracker!")
```
